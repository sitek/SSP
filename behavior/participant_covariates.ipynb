```python
import os

import pandas as pd
import numpy as np
```

```python
bids_dir = os.path.abspath('/bgfs/bchandrasekaran/krs228/data/SSP/data_bids/')
```

#### Load the `participants.tsv` file from the BIDS directory

```python
tsv_fpath = os.path.join(bids_dir, 'participants.tsv')
tsv_pd = pd.read_csv(tsv_fpath, delimiter='\t')
```

```python
tsv_pd.columns
```

```python
tsv_pd.sort_values(by=['participant_id'], inplace=True)
```

```python
tsv_pd.head()
```

```python
# make a set of Subject IDs that appear in tsv_pd
keep_ids = (tsv_pd['participant_id'].str.replace('sub-', '', regex=False).unique() )
print(keep_ids[:5])
```

#### Load the MRI subject log

```python
mri_table_fpath = 'SSP_MRI_2026-01-07.xlsx'

mri_table = pd.read_excel(mri_table_fpath)
```

```python
mri_table
```

```python
# filter mri_table
mri_table_filt = mri_table[mri_table['Subject'].isin(keep_ids)].reset_index(drop=True)
```

```python
mri_table_filt
```

#### Load the assessment covariate table

```python
cov_table_fpath = 'SSP_AUD_2026-01-09.xlsx'
cov_table = pd.read_excel(cov_table_fpath)
```

```python
cov_table
```

```python
# filter cov_table
cov_table_filt = cov_table[cov_table['Subject ID'].isin(keep_ids)].reset_index(drop=True)

#cov_table_filt.rename(columns={'Subject ID': 'participant_id'}, inplace=True)
```

```python
cov_table_filt
```

### Combine data into new participants_tsv table

```python
mri_cols_to_add = ['Subject', 'Age at MRI']
```

```python
cov_cols_to_add = [
    'Subject ID',
    'TONI_INDEX',
    'CTOPP_PhonologicalAwareness',
    'CELF_CoreLang',
    'CELF_Receptive',
    'CELF_ExpressiveLang',
    'PTA', 'win_correct_average', 'win_score_average' # not all participants
]

covar_rename = {
    'TONI_INDEX': 'toni_index',
    'CTOPP_PhonologicalAwareness': 'ctopp_phon_awareness',
    'CELF_CoreLang': 'celf_core_language',
    'CELF_Receptive': 'celf_receptive_language',
    'CELF_ExpressiveLang': 'celf_expressive_language',
    'PTA': 'pta',
    'win_correct_average': 'win_correct_avg',
    'win_score_average': 'win_score_avg'
}
```

```python
# add Subject ID to tsv_pd
tsv_pd = tsv_pd.assign(
    Subject_ID=tsv_pd['participant_id'].str.replace('sub-', '', regex=False) )

# merge selected columns from mri_table_filt
new_tsv_pd = tsv_pd.merge(
    mri_table_filt[mri_cols_to_add],
    left_on='Subject_ID',
    right_on='Subject',
    how='left'
)

# merge selected columns from cov_table_filt
new_tsv_pd = new_tsv_pd.merge(
    cov_table_filt[cov_cols_to_add],
    left_on='Subject_ID',
    right_on='Subject ID',
    how='left'
)

# replace the redundant subject ID columns
new_tsv_pd.drop(columns=['Subject_ID'], inplace=True)
new_tsv_pd.drop(columns=['Subject ID'], inplace=True)
new_tsv_pd.drop(columns=['Subject'], inplace=True)

# drop original 'age' (which was an estimate)
new_tsv_pd.drop(columns=['age'], inplace=True)
new_tsv_pd.rename(columns={'Age at MRI': 'age'}, inplace=True)

# BIDS-friendly renaming
new_tsv_pd.rename(columns=covar_rename, inplace=True)
new_tsv_pd['group'] = new_tsv_pd['group'].str.lower()
new_tsv_pd.replace({np.nan: 'n/a'}, inplace=True)
```

## Overwrite existing `participants.tsv` file

```python
new_tsv_pd.to_csv(
    tsv_fpath,
    sep='\t',
    index=False,
    na_rep='n/a'
)
```
